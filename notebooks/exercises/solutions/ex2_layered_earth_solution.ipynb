# Exercise 2: The layered earth (12 min)

Use the forward model to build intuition: what can TEM see, and what can it not tell apart?

Tip: you will compute and plot many models, so start by writing one short function that takes `thicknesses` and `resistivities`, computes the response and the apparent resistivity, and adds them to your plots.

### Functions you will need
```python
pytem.fwd_analytical_central(resistivity, tx_radius, times)          # half-space, closed form
pytem.fwd_circle_central(thicknesses, resistivities, tx_radius, times,
                         transform='dlf',            # or 'euler'
                         hankel_filter='key_101',    # or 'key_201'
                         fourier_filter='key_81')    # or 'key_101'
pytem.dbdt_to_apprho(dbdt, tx_moment, times)         # tx_moment = loop area x current [A m²]
pytem.plot_model(thicknesses, resistivities, ax=ax, label='...')
```
- `resistivities` has N entries, `thicknesses` N-1 (the last layer is the half-space). A half-space is `thicknesses=[]`, `resistivities=[rho]`.
- All forward functions return dB/dt in V/m² for 1 A. The step-off response is **negative**, so plot `-d` on log axes.

````
# pip install pytem
import time
import numpy as np
import matplotlib.pyplot as plt
import pytem

a = np.sqrt(40**2 / np.pi)        # 40 x 40 m loop [m]
times = np.logspace(-5, -2, 31)   # gate times [s]
````

### Task 2.1: conductor depth
A 10 m thick, 10 Ohm.m clay layer in a 100 Ohm.m sand (three layers). Put its top at 10, 30, 60, 100 and 150 m.

For each depth, plot -dB/dt against time and the apparent resistivity against time (two panels, one line per depth), and the models with `pytem.plot_model` in a third panel.

How does the time of the apparent-resistivity minimum change with depth? Below which depth does the 40 m loop stop seeing the clay?

````
def add(axs, thk, rho, a=a, times=times):
    d = -pytem.fwd_circle_central(thk, rho, a, times)
    axs[0].loglog(times, d, label=f'{thk} m, {rho} Ohm.m')
    axs[1].loglog(times, pytem.dbdt_to_apprho(d, np.pi * a**2, times))
    pytem.plot_model(thk, rho, ax=axs[2], label=None, depth_pad=50)
    return d

def labels(axs, title, zmax):
    axs[0].set(xlabel='Time [s]', ylabel='-dB/dt [V/m²]', title=title)
    axs[0].legend(fontsize=7)
    axs[1].set(xlabel='Time [s]', ylabel='Apparent resistivity [Ohm.m]')
    axs[2].set_ylim(zmax, 0)
    plt.tight_layout()

fig, axs = plt.subplots(1, 3, figsize=(14, 4))
for z in [10, 30, 60, 100, 150]:
    add(axs, [z, 10], [100, 10, 100])
labels(axs, 'Conductor depth', 210)
````

### Task 2.2: conductors vs resistors
Keep the layer at 30 m depth, 10 m thick, and set its resistivity to 1, 10, 1000 and 10000 Ohm.m. For each, print the maximum relative difference (%) from the response of a uniform 100 Ohm.m half-space.

Is TEM equally sensitive to a conductive and a resistive layer? Why?

````
d_hs = -pytem.fwd_circle_central([], [100], a, times)
for r in [1, 10, 1000, 10000]:
    d = -pytem.fwd_circle_central([30, 10], [100, r, 100], a, times)
    print(f'{r:>6} Ohm.m: max difference from half-space {np.max(np.abs(d - d_hs) / d_hs) * 100:6.1f} %')
````

### Task 2.3: equivalence
A thin conductor at 50 m depth in 100 Ohm.m. Use (thickness, resistivity) = (2 m, 2 Ohm.m), (4 m, 4 Ohm.m) and (8 m, 8 Ohm.m), so the conductance $S = h/\rho$ is the same. Plot the three responses and print the maximum relative difference of each from the 4 m case.

Compare the differences with a 3 % data error. What does this mean for an inversion?

````
fig, axs = plt.subplots(1, 3, figsize=(14, 4))
d = [add(axs, [50, h], [100, h, 100]) for h in [2, 4, 8]]
labels(axs, 'Equivalence, S = 0.5 S', 110)
for h, di in zip([2, 4, 8], d):
    print(f'h = rho = {h}: max difference from the 4 m case {np.max(np.abs(di - d[1]) / d[1]) * 100:.1f} %')
````

### Task 2.4 (if time): loop size
Repeat 2.1 with a 5 m and a 100 m loop radius (start `times` at 1e-6 s for the small loop). How does loop size change how deep you can see?

````
for ai, ti in [(5, np.logspace(-6, -3, 31)), (100, np.logspace(-5, -2, 31))]:
    fig, axs = plt.subplots(1, 3, figsize=(14, 4))
    for z in [10, 30, 60, 100, 150]:
        add(axs, [z, 10], [100, 10, 100], a=ai, times=ti)
    labels(axs, f'Loop radius {ai} m', 210)
````
